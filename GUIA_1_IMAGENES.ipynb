```python
import matplotlib.pyplot as plt
import numpy as np
from imageio import imread
from scipy import ndimage
```

```python
img=imread("onerice.bmp")
plt.imshow( img )
plt.show()
```

```python

def como_es(X):

    tamaño = X.shape

    mayor = np.max(X)

    menor = np.min(X)

    return tamaño, mayor, menor
```

```python
X = img[:,:]
print ( como_es ( X ) )
plt . imshow (X , cmap ='gray')
plt . show ()
```

```python
def segmenta(X, t):

    Y = np.zeros_like(X)
    Y[X > t] = 255

    area = np.sum(Y == 255)

    print("Área =", area)

    return Y
```

```python
umbral = [50,90 ,120, 180] #AQUI SE CAMBIA EL UMBRAL

for t in umbral:

    print("Umbral:", t)

    Y = segmenta(X, t)

    plt.imshow(Y, cmap="gray")

    plt.title(f"Segmentación con umbral {t}")

    plt.show()
```

```python
def calcula_centroide(Y):

    filas, columnas = np.where(Y == 255)

    fcentro = np.mean(filas)

    ccentro = np.mean(columnas)

    return fcentro, ccentro
```

```python
def dibuja_centroide(Y, filas, columnas):

    plt.imshow(Y, cmap="gray")

    plt.plot(columnas, filas, "ro", markersize=10)

    plt.title("Centroide")

    plt.show()
```

```python
Y = segmenta(X, 120) # aqui se cambia el umbral

fila, columna = calcula_centroide(Y)


print("Coordenadas del centroide:")
print("Fila:", fila)
print("Columna:", columna)


dibuja_centroide(Y, fila, columna)
```

```python
! wget https://www.dropbox.com/s/trcnlnrkoq9hcq3/rices.png
img2 = imread ("rices.png" )
plt . imshow ( img2 )
plt . show ()
R2 = img2 [: , : ] # Canal rojo
print ( como_es ( R2 ) )
plt . imshow ( R2 , cmap ="gray")
plt . show ()
```

```python
umbral = [50,90 ,120, 180] #AQUI SE CAMBIA EL UMBRAL

for t in umbral:

    print("Umbral:", t)

    Y1 = segmenta(R2, t)

    plt.imshow(Y1, cmap="gray")

    plt.title(f"Segmentación con umbral {t}")

    plt.show()
```

```python
(N, M) = R2.shape

Xm = np.zeros((N, M), dtype=np.uint8)

for i in range(N):
    xmin = np.min(R2[i, :])
    Xm[i, :] = R2[i, :] - xmin

plt.imshow(Xm, cmap="gray")
plt.show()
```

```python
Y2 = segmenta(Xm,50)
plt.imshow(Y2 ,cmap="gray")
plt.show()
```

```python
etiquetas, numero_granos = ndimage.label(Y2)
print("Número de granos de arroz:", numero_granos)
```
